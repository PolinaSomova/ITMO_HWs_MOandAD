## Сравнительная таблица метрик классификации

| Метрика | Интерпретация | Диапазон | Когда использовать | Преимущества | Недостатки | Особенности |
|---------|---------------|----------|-------------------|--------------|-----------|-------------|
| **Accuracy** (Точность) | Доля правильных предсказаний | [0, 1] | Сбалансированные данные; все классы одинаково важны | Интуитивна; просто считается; легко объяснить | **Вводит в заблуждение** при дисбалансе; не видит разницу между типами ошибок; модель может быть 95% точной, предсказав всё как класс большинства | ✗ Не рекомендуется как основная метрика при дисбалансе |
| **Precision** (Точность предсказаний) | Из предсказанных позитивов, сколько было правильных | [0, 1] | Когда FP дорого; спам-фильтры, fraud-детекция, модерация контента | Показывает надёжность положительных предсказаний; низкий FP-rate | Игнорирует FN (пропущенные случаи); может быть высокой при очень консервативной модели | «Из всех писем, отмеченных как спам, сколько действительно спам?» |
| **Recall** (Полнота)  | Из реальных позитивов, сколько модель нашла | [0, 1] | Когда FN дорого; медицина (диагностика), безопасность, обнаружение мошенничества | Показывает способность находить позитивы; низкий FN-rate | Игнорирует FP (ложные срабатывания); может быть высокой при очень либеральной модели | «Из всех больных людей, сколько модель правильно диагностировала?» |
| **F1-Score**  | Гармоническое среднее precision и recall | [0, 1] | Дисбалансированные данные; когда важны обе ошибки; **стандартная метрика** | Балансирует precision и recall; штрафует экстремумы; хороша для имбалансированных данных | Не настраивается; одинаков вес для precision и recall; не учитывает TN | Становится 0, если precision ИЛИ recall = 0 (строго) |
| **F-Beta Score** | Гибридная F-score с регулируемым весом | [0, 1] | Когда нужен баланс, но precision или recall важнее; β>1 → recall, β<1 → precision | Настраивается под задачу; лучше, чем F1 для асимметричных затрат | Нужно выбирать β; менее интерпретируема; может быть непонятна stakeholders | β=1 = F1; β=2 даёт recall вес ≈2x; β=0.5 даёт precision вес ≈2x |
| **ROC-AUC** | Способность модели разделить классы на всех порогах | [0, 1] | Сбалансированные данные; нужна оценка на всех порогах; кредиты, медицина | Инвариантна к порогу; показывает ранжирование; стандартная в индустрии | **Вводит в заблуждение при дисбалансе**; не учитывает precision; может быть 0.9 при 99% accuracy с нулевой precision | ROC-AUC=0.5 → random guessing; AUC=1 → perfect ranking |
| **MCC** (Matthews) |Корреляция между предсказаниями и реальностью | [-1, 1] | **Рекомендуется при дисбалансе** вместо ROC-AUC; бинарная классификация; реальные данные | Не зависит от дисбаланса; учитывает все 4 типа ошибок; высокий MCC = надёжная модель | Сложнее объяснить; -1 до +1 (нужен shift в мышлении); менее известна | MCC=0 → не лучше random; MCC=1 → perfect; MCC=-1 → opposite |

### Бизнес-постановка задачи



Для банка удержание существующего клиента обходится в 5–7 раз дешевле, чем привлечение нового [ссылка] [reference:0]. При среднем уровне оттока 25–30% ежегодно [ссылка] [reference:1] банк теряет значительную часть прибыли.

**Цель:**
Разработать модель, которая по поведенческим и демографическим данным клиента предскажет вероятность его ухода.

## 2. Выбор метрик

В первую очередь важно обозначит, что перед нами стоит задача классификации. В EDA был проведен анализ на соотношение ушедших и оставших  клиентов: клиенты, которые ушли, значительно меньше, чем тех, кто остался. То есть данные диссбалансированы. В такой ситуации метрика **Accuracy** (доля правильных ответов) становится неинформативной. Модель, предсказывающая для всех клиентов «0», покажет высокую точность (около 80%), но будет бесполезна на практике, так как не найдёт ни одного уходящего клиента. А по задаче критически важно находить именно тех, кто собирается уйти (например, чтобы предложить им удерживающие акции). При этом также важно минимзировать ошибки, чтобы не тратить бюджет на лояльных клиентов (готовых остатьатья и так).

Таким образом, налиучшим выбором будет метрика: **F1-Score**, где **Precision** - это точность (из всех предсказанных уходов сколько реальных), а **Recall** - это полнота (из всех реальных уходов сколько). *F1-Score - гармоническое среднее precision и recall [0, 1], когда дисбалансированные данные и важны обе ошибки*

Дополнительно необходимо отслеживать **Precision** и **Recall** по отдельности, чтобы понимать, какой тип ошибок преобладает. При необходимости можно будет скорректировать порог классификации в пользу Recall (чтобы не пропускать уходящих) или Precision (чтобы не тратить бюджет на лишние акции).

### Анализ распределения числовых признаков в разрезе оттока

```python
numeric_features = ['CreditScore', 'Age', 'Tenure', 'Balance', 'NumOfProducts', 'EstimatedSalary']
fig, axes = plt.subplots(2, 3, figsize=(14, 9))
axes = axes.flatten()

for i, feature in enumerate(numeric_features):
    sns.histplot(data=df, x=feature, hue='Exited', kde=True, ax=axes[i], palette='Set1')
    axes[i].set_title(f'Распределение {feature}')

plt.tight_layout()
plt.show()
```

Вывод по призанкам:

* **Age**: Наиболее сильное различие. У ушедших клиентов распределение сдвинуто вправо — пик приходится на 30–50 лет. У лояльных — широкое плато с пиком около 35 лет. Возраст — важный предиктор.
* **Tenure**: Стаж почти одинаков в обоих классах, различие минимально.
* **Balance**: У ушедших заметно больше клиентов с очень низким балансом (около 0) и, парадоксально, с высоким балансом.
* **NumOfProducts**: Клиенты с 1 продуктом чаще остаются, с 2 — тоже. С 3–4 продуктами доля оттока резко возрастает. Предположение: возможно, сложность обслуживания или недовольство.
* **EstimatedSalary**: Распределения почти идентичны, зарплата сама по себе слабо связана с оттоком.

Вывод для предобработки (дополнительно)
* Признаки `Tenure` и `EstimatedSalary`, вероятно, можно оставить без изменений — их вклад будет мал.
* Возраст (`Age`) и количество продуктов (`NumOfProducts`) — ключевые признаки для модели.

### Анализ категориальных признаков

```python
categorical_features = ['Geography', 'Gender', 'HasCrCard', 'IsActiveMember']

fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.flatten()

for i, feature in enumerate(categorical_features):
    cross_tab = pd.crosstab(df[feature], df['Exited'], normalize='index') * 100
    ax = cross_tab.plot(kind='bar', ax=axes[i], color=['#1f77b4', '#ff7f0e'])
    ax.set_title(f'Доли по признаку {feature}')
    ax.set_ylabel('Доля, %')
    ax.set_xlabel(feature)
    ax.legend(['Остался', 'Ушёл'])
    ax.tick_params(axis='x', rotation=45)

    # Подписываем оба столбца
    for p in ax.patches:
        height = p.get_height()
        if not np.isnan(height):
            ax.text(p.get_x() + p.get_width()/2., height + 0.5,
                    f'{height:.1f}%', ha='center', va='bottom', fontsize=8)

plt.tight_layout()
plt.show()
```

Вывод по категориальным признакам:

* **Geography**: Германия выделяется доля оттока ~32%, тогда как во Франции и Испании ~16%. Предполложение: данный признак важен для модели.
* **Gender**: Женщины уходят заметно чаще мужчин (~25% против ~16%).
*  **HasCrCard**: Наличие кредитной карты почти не влияет на отток. Предположение: признак можно исключить.
*  **IsActiveMember**: Активные клиенты уходят реже (активные: ~14%, неактивные: ~27%). Важный предиктор.

Как кодировать признаки:

* `Geography` и `Gender` - **One-Hot Encoding**.
